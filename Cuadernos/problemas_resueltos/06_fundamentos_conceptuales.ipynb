# Problema resuelto 06: fundamentos conceptuales

## Problema

Explicar por que una teoria cuantica relativista basada solo en particulas con numero fijo no es suficiente y por que la descripcion en terminos de campos resuelve el problema.


## Paso 1. El conflicto inicial

En mecanica cuantica no relativista es razonable trabajar con un espacio de Hilbert de una particula o de un numero fijo de particulas. Sin embargo, al imponer compatibilidad con relatividad especial aparece la posibilidad de convertir energia cinetica en masa:

$$
E^2 = \mathbf p^2 + m^2.
$$

Si la energia disponible es suficientemente alta, procesos de creacion y aniquilacion dejan de ser excepcionales y pasan a ser inevitables.


## Paso 2. Que falla en la descripcion de particulas fijas

Un espacio de Hilbert con numero fijo de particulas no acomoda naturalmente transiciones como

$$
\gamma \to e^- + e^+,
$$

ni tampoco la descripcion coherente de procesos inversos. La teoria tendria que introducir de manera externa cambios de sector que, en QFT, quedan incorporados desde el principio en el espacio de Fock.


## Paso 3. Solucion conceptual

La QFT toma a los campos como entidades fundamentales y a las particulas como excitaciones cuantizadas de esos campos. Asi, el vacio, los estados de una particula y los estados multiparticle forman parte de una unica estructura algebraica.

El operador de creacion $a^\dagger$ anade un cuanto y el operador de aniquilacion $a$ lo elimina. Esta es la manera mas natural de describir procesos relativistas.

## Conclusion

La necesidad de QFT no nace de una preferencia formal, sino de una exigencia fisica: una teoria cuantica compatible con relatividad debe permitir creacion y destruccion de particulas. El lenguaje de campos cumple esa exigencia sin renunciar a localidad, causalidad y simetria relativista.
