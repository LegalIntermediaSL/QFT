# Problema resuelto 06: Fundamentos conceptuales

Notebook asociado a `Tutorial/01_fundamentos_conceptuales/README.md`.

## Problema

Explicar por que la QFT necesita simultaneamente:

- relatividad especial;
- cuantizacion;
- localidad;
- numero variable de particulas.

## Solucion orientativa

La relatividad impone estructura causal y covariancia. La cuantizacion introduce amplitudes y operadores. La localidad protege la causalidad microfisica. La equivalencia entre energia y masa obliga a permitir creacion y destruccion de particulas. Los campos cuanticos son el lenguaje que acomoda todas estas exigencias a la vez.
