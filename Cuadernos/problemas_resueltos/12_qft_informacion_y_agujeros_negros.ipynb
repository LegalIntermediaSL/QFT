# Problema resuelto 12: QFT, informacion y agujeros negros

## Problema

Explicar por que la radiacion de Hawking puede describirse como un fenomeno de QFT en espacio-tiempo curvo y por que eso conduce a la paradoja de la informacion.


## Paso 1. El vacio depende del observador

En espacio-tiempo curvo o en presencia de horizontes, la nocion de particula depende de como se definan modos positivos y negativos. Lo que un observador interpreta como vacio puede no ser vacio para otro.


## Paso 2. Produccion efectiva de particulas

Cerca del horizonte de un agujero negro, la estructura de modos produce un flujo termico efectivo hacia el infinito con temperatura de Hawking

$$
T_H = \frac{\kappa}{2\pi},
$$

donde $\kappa$ es la gravedad superficial en unidades naturales.


```python
import numpy as np

kappa = 0.2
T_H = kappa / (2 * np.pi)
T_H
```

## Paso 3. Donde aparece la paradoja

Si la evaporacion completa deja solo radiacion termica aparentemente mixta, una evolucion que empezo en un estado puro parece terminar en un estado mixto. Eso choca con la unitaridad de la mecanica cuantica.

La tension entre semiclasica, horizonte y unitaridad es el nucleo de la paradoja de la informacion.


## Conclusion

La discusion sobre agujeros negros no es un apendice exotico: obliga a examinar que entendemos por estado cuantico, observador, entropia y localidad. Por eso este tema conecta QFT, gravedad y teoria de la informacion en una misma frontera conceptual.
