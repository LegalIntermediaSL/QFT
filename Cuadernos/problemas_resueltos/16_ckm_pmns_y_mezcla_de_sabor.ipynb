# Problema resuelto 16: CKM, PMNS y mezcla de sabor

## Problema

Entender por que la mezcla de sabor aparece al diagonalizar matrices de masa y como se refleja de forma distinta en el sector quark y en el sector leptónico.

## Paso 1. Dos bases distintas

La mezcla de sabor nace cuando la base natural de interaccion gauge no coincide con la base que diagonaliza las matrices de masa. Esta desalineacion es el origen comun de CKM y PMNS.

## Paso 2. Caso quark y caso leptónico

En quarks, la matriz CKM aparece en corrientes cargadas. En leptones, la matriz PMNS se vuelve visible sobre todo en oscilaciones de neutrinos. La idea estructural es la misma, aunque la fenomenologia experimental cambia.

```python
import numpy as np

theta = 0.22
V_demo = np.array([
    [np.cos(theta), np.sin(theta)],
    [-np.sin(theta), np.cos(theta)]
])
np.round(V_demo, 4)
```

## Paso 3. Lectura fisica

La matriz de ejemplo anterior no es una CKM o PMNS realista completa, pero ayuda a fijar la idea de mezcla: una corriente cargada puede conectar un estado con combinaciones lineales de otros sabores, no con un unico estado puro en la base de masa.
