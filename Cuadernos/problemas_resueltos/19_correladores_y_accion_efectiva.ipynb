# Problema resuelto 19: correladores y accion efectiva

## Problema

Seguir de forma guiada la cadena conceptual que va desde el funcional generador $Z[J]$ hasta la accion efectiva $\Gamma[\phi_c]$ y el potencial efectivo.

## Resultado esperado

Recorrer la cadena entre funcional generador, correladores, campo clasico y accion efectiva.


## Paso 1. Funcional generador y correladores

El formalismo funcional organiza correladores a partir de derivadas respecto de la fuente externa:

$$
\langle 0|T\phi(x_1)\cdots\phi(x_n)|0\rangle = \left.\frac{1}{i^n}\frac{\delta^n Z[J]}{\delta J(x_1)\cdots\delta J(x_n)}\right|_{J=0}.
$$

Eso convierte a $Z[J]$ en una especie de generador compacto de correladores.

```python
Z_demo = 'exp(i/2 J Delta J)'
correladores = {
    '1_punto': 'delta Z / delta J',
    '2_puntos': 'delta^2 Z / delta J^2',
    'n_puntos': 'delta^n Z / delta J^n'
}
{'Z_demo': Z_demo, 'correladores': correladores}
```

## Paso 2. De W[J] a Gamma[phi_c]

Definimos primero

$$
W[J] = -i \log Z[J],
$$

y luego el campo clasico esperado

$$
\phi_c = \frac{\delta W[J]}{\delta J}.
$$

La accion efectiva surge por transformacion de Legendre:

$$
\Gamma[\phi_c] = W[J] - \int d^4x\, J\phi_c.
$$


```python
phi_c_demo = [-1.0, 0.0, 1.0]
Gamma_demo = [0.6, 0.0, 0.6]
{'phi_c_demo': phi_c_demo, 'Gamma_demo': Gamma_demo, 'lectura': 'minimos de Gamma sugieren vacios efectivos'}
```
