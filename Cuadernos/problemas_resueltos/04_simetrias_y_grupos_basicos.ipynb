## Problema 1

Explicar por que una rotacion por un angulo $\theta_1$ seguida de otra por $\theta_2$ sigue siendo una rotacion del mismo tipo.

La composicion produce una nueva rotacion por angulo $\theta_1+\theta_2$. Esto ilustra el cierre del grupo de rotaciones.

## Problema 2

Interpretar la transformacion global de fase

$$\phi \to e^{i\alpha}\phi$$

como una simetria continua.

La familia de transformaciones esta parametrizada continuamente por $\alpha$. Para valores pequeños,

$$e^{i\alpha} \approx 1 + i\alpha,$$

lo que muestra el papel del generador infinitesimal.

## Problema 3

Explica por que las simetrias son utiles para construir una teoria y no solo para describirla.

Porque restringen que terminos pueden aparecer en el lagrangiano, que cantidades se conservan y como transforman los campos. En QFT moderna, las simetrias guian gran parte de la construccion teorica.
