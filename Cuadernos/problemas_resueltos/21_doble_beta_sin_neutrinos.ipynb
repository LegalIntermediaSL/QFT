# Problema resuelto 21: doble beta sin neutrinos

## Problema

Fijar la relacion entre violacion de numero leptónico, neutrinos de Majorana y el observable efectivo de doble beta sin neutrinos.

## Resultado esperado

Relacionar violacion de numero leptonico, masas de Majorana y el observable efectivo de doble beta sin neutrinos.


## Paso 1. Cambio de numero leptónico

En la desintegracion doble beta sin neutrinos el proceso esquematico es

$$
(A,Z) \to (A,Z+2) + 2e^-.
$$

No hay neutrinos salientes, de modo que el numero leptónico cambia en dos unidades: $\Delta L = 2$.

```python
leptones_iniciales = 0
leptones_finales = 2
delta_L = leptones_finales - leptones_iniciales
{'L_inicial': leptones_iniciales, 'L_final': leptones_finales, 'Delta_L': delta_L}
```

## Paso 2. Masa efectiva de Majorana

En la lectura mas simple, la amplitud depende de

$$
m_{\beta\beta} = \left|\sum_i U_{ei}^2 m_i\right|.
$$

Eso muestra que el observable combina masas, mezcla leptónica y fases complejas.

```python
m_i = [0.01, 0.012, 0.05]
Uei_sq_demo = [0.67, 0.29, 0.02]
m_beta_beta_demo = sum(u*m for u, m in zip(Uei_sq_demo, m_i))
{'m_i_eV_demo': m_i, 'Uei_sq_demo': Uei_sq_demo, 'm_beta_beta_demo_eV': round(m_beta_beta_demo, 6)}
```
