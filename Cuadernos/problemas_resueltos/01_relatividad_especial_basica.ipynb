## Objetivo

Resolver tres tipos de ejercicios basicos:

1. clasificacion de intervalos relativistas;
2. calculo de la norma de un cuatro-vector;
3. verificacion de la relacion de dispersion relativista.

## Convencion

Usamos la metrica:

$$\eta_{\mu\nu}=\mathrm{diag}(1,-1,-1,-1)$$

y unidades naturales $c=1$.

```python
from math import sqrt

def intervalo(t, x, y=0.0, z=0.0):
    return t**2 - x**2 - y**2 - z**2

def clasificar_intervalo(s2):
    if s2 > 0:
        return 'tipo tiempo'
    if s2 < 0:
        return 'tipo espacio'
    return 'tipo luz'
```

## Problema 1

Clasificar la separacion entre los eventos:

$$x^\mu=(5,3,0,0).$$

```python
s2 = intervalo(5, 3)
print('s^2 =', s2)
print('clasificacion =', clasificar_intervalo(s2))
```

### Resultado

Como $s^2=25-9=16>0$, la separacion es **de tipo tiempo**. Eso significa que, en principio, un evento podria influir causalmente sobre el otro.

## Problema 2

Tomar el cuatro-momento

$$p^\mu=(10,6,8,0)$$

y calcular su masa invariante usando:

$$p_\mu p^\mu = E^2-\mathbf{p}^2.$$

```python
E, px, py, pz = 10, 6, 8, 0
m2 = E**2 - px**2 - py**2 - pz**2
m = sqrt(m2)
print('m^2 =', m2)
print('m =', m)
```

### Resultado

Se obtiene $m^2=0$, luego el sistema es **sin masa**. Este es exactamente el tipo de condicion que satisface una particula ultrarrelativista como el foton.

## Problema 3

Verificar la relacion de dispersion para una particula de masa $m=2$ y momento espacial $|\mathbf{p}|=3$.

```python
m = 2
p = 3
E = sqrt(p**2 + m**2)
print('E =', E)
print('Comprobacion E^2 - p^2 =', E**2 - p**2)
```

### Interpretacion fisica

La igualdad $E^2-p^2=m^2$ muestra que la masa es un invariante relativista del sistema. En QFT, esta misma relacion reaparece como condicion on-shell para los modos libres de un campo.
