# Problema resuelto 08: accion y Noether

## Problema

Mostrar como una simetria continua del lagrangiano genera una corriente conservada.


## Paso 1. Lagrangiano del campo escalar complejo

$$
\mathcal L = \partial_\mu \phi^* \, \partial^\mu \phi - m^2 \phi^* \phi.
$$

Este lagrangiano es invariante bajo la transformacion global $U(1)$

$$
\phi \to e^{i\alpha} \phi, \qquad \phi^* \to e^{-i\alpha} \phi^*.
$$


## Paso 2. Corriente de Noether

La corriente asociada es

$$
j^\mu = i \left( \phi^* \partial^\mu \phi - (\partial^\mu \phi^*) \phi \right),
$$

y satisface

$$
\partial_\mu j^\mu = 0.
$$


```python
import sympy as sp

alpha = sp.symbols('alpha', real=True)
sp.series(sp.exp(sp.I * alpha), alpha, 0, 2)
```

La expansion a primer orden recuerda que la variacion infinitesimal es $\delta \phi = i \alpha \phi$. A partir de ahi se obtiene la corriente conservada.


## Conclusion

El mensaje estructural es profundo: las cantidades conservadas no se postulan por separado, sino que nacen de simetrias continuas del principio de accion. Esta idea atraviesa toda la fisica moderna, desde la conservacion de carga hasta las teorias gauge.
