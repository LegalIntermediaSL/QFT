# Problema resuelto 08: Accion y Noether

Notebook asociado a `Tutorial/03_accion_y_simetrias/README.md`.

## Problema

Explicar como se pasa de una accion invariante bajo una simetria continua a una cantidad conservada.

## Resolucion conceptual

La invariancia infinitesimal de la accion permite reorganizar la variacion en forma de divergencia de una corriente. Esa corriente satisface una ecuacion de conservacion y, al integrar su componente temporal, se obtiene una carga conservada. Esta es la esencia del teorema de Noether.
