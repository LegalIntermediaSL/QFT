## Paso 2. Descomposicion del lagrangiano

$$
\mathcal L_{\text{SM}} = \mathcal L_{\text{gauge}} + \mathcal L_{\text{ferm}} + \mathcal L_{\text{Higgs}} + \mathcal L_{\text{Yukawa}}.
$$

- El sector gauge describe bosones portadores de fuerza.
- El sector fermionico contiene quarks y leptones.
- El sector de Higgs implementa la ruptura espontanea de simetria.
- El sector de Yukawa genera masas fermionicas y mezclas.
