# Problema resuelto 11: Estructura del Modelo Estandar

Notebook asociado a `Tutorial/10_modelo_estandar/01_lagrangiano_del_modelo_estandar.md`.

## Problema

Describir el Lagrangiano del Modelo Estandar como suma de bloques fisicos.

## Resolucion

El esquema mas compacto es:

$$\mathcal{L}_{\text{SM}} = \mathcal{L}_{\text{gauge}} + \mathcal{L}_{\text{ferm}} + \mathcal{L}_{\text{Higgs}} + \mathcal{L}_{\text{Yukawa}}.$$ 

Cada bloque cumple una funcion distinta: propagacion de bosones gauge, dinamica de fermiones, ruptura electrodébil y generacion de masas fermionicas.
