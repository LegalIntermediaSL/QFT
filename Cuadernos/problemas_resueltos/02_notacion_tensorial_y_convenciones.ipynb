## Problema 1

Dado el cuatro-vector

$$x^\mu=(4,1,2,3),$$

bajar el indice con la metrica $(+,-,-,-)$.

```python
x_up = (4, 1, 2, 3)
eta = (1, -1, -1, -1)
x_down = tuple(e * x for e, x in zip(eta, x_up))
print('x_mu =', x_down)
```

### Resultado

Se obtiene:

$$x_\mu=(4,-1,-2,-3).$$

## Problema 2

Calcular el producto escalar relativista entre

$$a^\mu=(2,1,0,0), \qquad b^\mu=(3,1,1,0).$$

```python
a = (2, 1, 0, 0)
b = (3, 1, 1, 0)
producto = a[0]*b[0] - a[1]*b[1] - a[2]*b[2] - a[3]*b[3]
print('a·b =', producto)
```

### Interpretacion

El producto relativista no coincide con el producto euclideo usual. La parte espacial entra con signo negativo por la estructura de Minkowski.
