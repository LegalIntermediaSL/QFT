# Problema resuelto 15: regularizacion dimensional y running

## Problema

Entender por que la regularizacion dimensional introduce polos en $1/\varepsilon$, como aparece la escala $\mu$ y por que una funcion beta resume la dependencia con la escala.

## Paso 1. Regularizar antes de renormalizar

En una integral de lazo ultravioleta divergente, la regularizacion dimensional reemplaza la dimension fisica por una expresion del tipo $d = 4 - \varepsilon$. Eso convierte la divergencia en un polo algebraico manejable.

## Paso 2. Polo en $1/\varepsilon$ y escala $\mu$

La escala $\mu$ mantiene consistencia dimensional en el calculo y el polo en $1/\varepsilon$ señala la parte aun no renormalizada. La cuenta no termina ahi: hace falta absorber esa dependencia en contraterminos o en parametros renormalizados.

```python
import numpy as np

eps = np.array([0.5, 0.2, 0.1, 0.05])
poles = 1.0 / eps
mu_values = np.array([1.0, 2.0, 5.0])
logs = np.log(mu_values)

{'poles_demo': list(zip(eps.tolist(), np.round(poles, 3).tolist())), 'log_mu_demo': list(zip(mu_values.tolist(), np.round(logs, 4).tolist()))}
```

## Paso 3. Running y funcion beta

Una vez renormalizada la teoria, la dependencia con la escala no desaparece del todo: se reorganiza en el running de acoplamientos y masas. La funcion beta resume precisamente como cambia un acoplamiento al cambiar la escala de renormalizacion.

```python
mu = np.array([1.0, 2.0, 5.0, 10.0])
g0 = 0.3
b = 0.02
g_running_demo = g0 / (1 - b * np.log(mu))
list(zip(mu.tolist(), np.round(g_running_demo, 4).tolist()))
```
