# Problema resuelto 14: neutrinos y oscilaciones

## Problema

Entender de forma guiada por que los estados de sabor y los estados de masa de neutrinos no coinciden en general y como esa desalineacion conduce a oscilaciones.

## Paso 1. Dos bases distintas

La idea central es distinguir entre:

- estados de sabor, que son los que participan naturalmente en interacciones debiles;
- estados de masa, que son los que evolucionan con fases bien definidas.

Si ambas bases no coinciden, aparece la mezcla leptónica.

## Paso 2. Mezcla y fases relativas

En una descripcion minima de dos sabores, basta imaginar una matriz de mezcla sencilla y dos masas distintas. La evolucion acumula fases relativas diferentes para cada estado de masa, y al recombinarse produce una probabilidad de deteccion dependiente de la distancia o del tiempo.

```python
import numpy as np

theta = 0.6
dm2 = 7.5e-5
L_over_E = np.linspace(0, 20000, 6)
prob = np.sin(2 * theta) ** 2 * np.sin(1.27 * dm2 * L_over_E) ** 2

list(zip(np.round(L_over_E, 1), np.round(prob, 4)))
```

## Paso 3. Lectura fisica

La expresion numerica anterior no pretende ser una simulacion realista completa, pero si fijar la intuicion correcta: la probabilidad de cambiar de sabor oscila porque los estados de masa no acumulan la misma fase.

Ese comportamiento no seria posible si los neutrinos fueran estrictamente masivos cero y sin mezcla.

## Paso 4. Conexion con PMNS

En el caso real de tres sabores, la matriz PMNS organiza la mezcla leptónica. La moraleja conceptual, sin embargo, ya aparece en el ejemplo simple: la oscilacion es la huella observable de la desalineacion entre base de sabor y base de masa.
