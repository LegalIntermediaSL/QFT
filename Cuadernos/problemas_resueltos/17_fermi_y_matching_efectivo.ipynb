# Problema resuelto 17: Fermi y matching efectivo

## Objetivo

Seguir de forma guiada el paso entre una teoria mediada por un boson pesado y la descripcion efectiva local de cuatro fermiones que aparece en la teoria de Fermi.

## Resultado esperado

Poder leer el paso entre una teoria mediada por un boson pesado y su operador efectivo dominante a baja energia.


A bajas energias, el propagador de un boson pesado puede expandirse como

$$
\frac{1}{q^2-M^2} \approx -\frac{1}{M^2}\left(1 + \frac{q^2}{M^2} + \cdots\right),
$$

siempre que $q^2 \ll M^2$. El primer termino produce un operador local efectivo.

```python
import numpy as np

g = 0.65
M_W = 80.4
GF_estimado = g**2 / (8 * M_W**2)
GF_estimado
```

## Lectura fisica

La cantidad anterior ilustra la estructura de escala correcta: el coeficiente efectivo hereda dos potencias del acoplamiento y dos potencias inversas de la masa pesada. Eso explica por que la teoria de Fermi captura muy bien la fisica debil a energias mucho menores que $M_W$.

```python
q = np.array([1.0, 5.0, 20.0, 50.0])
aprox = -1 / M_W**2
exacto = 1 / (q**2 - M_W**2)
comparacion = np.round((exacto - aprox) / aprox, 6)
{'q_GeV': q.tolist(), 'error_relativo': comparacion.tolist()}
```

## Advertencia

Esta comparacion numerica no reemplaza un matching completo entre amplitudes, corrientes y normalizaciones espinoriales. Sirve para fijar la idea esencial: cuando $q^2/M_W^2$ es pequeño, la descripcion efectiva local es la aproximacion natural.
