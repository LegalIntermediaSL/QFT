## Problema 1

Dado el espectro

$$E_n = \omega\left(n+\frac{1}{2}\right),$$

calcular los primeros cinco niveles para $\omega=2$.

```python
omega = 2
niveles = [omega*(n + 0.5) for n in range(5)]
niveles
```

## Problema 2

Verificar la separacion uniforme entre niveles.

```python
diferencias = [niveles[i+1] - niveles[i] for i in range(len(niveles)-1)]
diferencias
```

### Interpretacion

La separacion constante entre niveles es exactamente lo que hace del oscilador armonico el modelo algebraico ideal para entender los cuantos de excitacion de un campo libre.

## Problema 3

Explica por que el termino $\omega/2$ se interpreta como energia de punto cero.

El estado fundamental corresponde a $n=0$, pero incluso ahi queda:

$$E_0 = \frac{\omega}{2}. $$

Por eso el vacio del oscilador no coincide con energia nula. Esta es la semilla conceptual de la energia de vacio en QFT.
