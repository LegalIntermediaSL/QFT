# Problema resuelto 18: corriente de Dirac y limite no relativista

## Problema

Entender como la corriente de Dirac resume una cantidad conservada y por que el limite no relativista conecta la teoria de Dirac con la intuicion de Pauli y Schroedinger.

## Resultado esperado

Seguir la corriente conservada de Dirac y reconocer el sentido fisico del limite no relativista.


## Paso 1. Corriente conservada

Para el lagrangiano de Dirac, la corriente asociada a la simetria global $U(1)$ se escribe como

$$
j^\mu = \bar{\psi}\gamma^\mu\psi.
$$

La componente temporal $j^0 = \psi^\dagger \psi$ se interpreta como densidad de probabilidad o densidad de carga, segun el contexto.

```python
psi = [1+0j, 0.2+0.1j, 0.05+0j, 0.01+0j]
densidad = sum(abs(z)**2 for z in psi)
{'psi_componentes_demo': [str(z) for z in psi], 'j0_demo': round(densidad, 6)}
```

## Paso 2. Separacion entre componentes grandes y pequeñas

En el limite no relativista, la masa domina la escala dinamica y las componentes inferiores del espinor quedan suprimidas. Eso permite reorganizar la ecuacion de Dirac en una dinamica efectiva cercana a la ecuacion de Pauli.

```python
p_sobre_m = [0.01, 0.05, 0.1]
supresion_componentes = [round(x, 4) for x in p_sobre_m]
{'p_sobre_m': p_sobre_m, 'tamano_relativo_componentes_pequenas_demo': supresion_componentes}
```

## Advertencia

Este cuaderno fija la intuicion del paso relativista al no relativista, pero no sustituye una derivacion completa de Foldy-Wouthuysen ni del limite de Pauli. La idea que queremos conservar es mas simple: cuando $|\vec{p}| \ll m$, la estructura espinorial se reorganiza de forma controlada.
