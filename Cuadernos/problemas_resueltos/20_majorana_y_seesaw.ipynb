## Problema

Seguir de forma guiada la intuicion minima del mecanismo seesaw y su conexion con masas de Majorana para neutrinos.

## Resultado esperado

Seguir la intuicion matricial del seesaw y relacionarla con masas pequenas de neutrinos.


## Paso 1. Matriz de masa minima

En el esquema mas simple, la mezcla entre un estado ligero y un neutrino derecho pesado se resume en

$$
M = \begin{pmatrix} 0 & m_D \\ m_D & M \end{pmatrix},
$$

con $M \gg m_D$.

```python
mD = 100.0
M = 1e14
m_ligera_demo = mD**2 / M
{'mD_GeV': mD, 'M_GeV': M, 'm_ligera_demo_GeV': m_ligera_demo}
```

## Paso 2. Lectura fisica

La formula $m_\nu \sim m_D^2/M$ resume la intuicion del seesaw: una escala pesada muy grande empuja el autovalor ligero hacia abajo. De este modo, una masa neutrínica pequeña deja de parecer arbitraria y pasa a interpretarse como efecto de una jerarquia de escalas.

```python
M_vals = [1e8, 1e10, 1e12, 1e14]
masas = [mD**2 / x for x in M_vals]
{'M_vals_GeV': M_vals, 'm_ligera_GeV': masas}
```
