# Problema resuelto 10: Interacciones y perturbaciones

Notebook asociado a `Tutorial/05_interacciones_y_perturbaciones/README.md`.

## Problema

Explicar por que un diagrama de Feynman no es una trayectoria clasica sino un termino de una expansion perturbativa.

## Resolucion

La matriz S se expande en la serie de Dyson y, al reorganizar productos temporales de campos, aparecen contracciones que se representan graficamente. Cada diagrama codifica propagadores, vertices y orden en el acoplamiento. El significado fisico final pertenece a la suma de todas las contribuciones relevantes, no a un dibujo aislado.
