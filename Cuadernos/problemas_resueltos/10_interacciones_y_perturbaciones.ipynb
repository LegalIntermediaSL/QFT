# Problema resuelto 10: interacciones y perturbaciones

## Problema

Entender por que la matriz $S$ se expande perturbativamente y como esa expansion conduce a reglas diagramaticas.


## Paso 1. Separacion entre parte libre e interaccion

Escribimos el Hamiltoniano como

$$
H = H_0 + H_{\text{int}}.
$$

La parte libre se resuelve exactamente, mientras que la interaccion se trata como expansion en una constante de acoplamiento pequena.


## Paso 2. Serie de Dyson

Formalmente,

$$
S = T \exp\left[-i \int d^4 x \, \mathcal H_{\text{int}}(x)\right].
$$

Al expandir la exponencial aparece una serie ordenada temporalmente. Cada termino tiene una interpretacion combinatoria que los diagramas de Feynman vuelven visualmente manejable.


```python
acoplamiento = 0.1
ordenes = [acoplamiento**n for n in range(1, 5)]
ordenes
```

## Paso 3. Significado fisico

Si el acoplamiento es pequeno, los terminos de orden superior suelen estar suprimidos. Por eso la expansion perturbativa puede ser util en QED y menos fiable en contextos de acoplamiento fuerte.


## Conclusion

La teoria de perturbaciones no es exacta, pero organiza calculos de forma sistematica. Los diagramas no reemplazan el formalismo: lo condensan.
