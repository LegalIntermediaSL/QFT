# Problema resuelto 13: gauge-fixing y scattering en QED

## Problema

Entender por que la fijacion de gauge es necesaria en QED y como se organiza una amplitud elemental de scattering con propagador del foton y vertices fermion-foton.

## Paso 1. Simetria gauge y redundancia

El potencial electromagnetico contiene redundancia gauge. Para definir limpiamente el propagador del foton y avanzar en calculos perturbativos, conviene añadir un termino de gauge-fixing adecuado. Esto no cambia la fisica observable, pero si ordena el formalismo.

## Paso 2. Ingredientes minimos de una amplitud en QED

En un scattering elemental mediado por intercambio de foton aparecen:

- un vertice fermion-foton en cada extremo;
- un propagador del foton en la linea interna;
- espinores externos para estados de entrada y salida.

La estructura analitica ya deja ver que una amplitud no es una trayectoria, sino una composicion de factores locales.

```python
e = 0.3
q2 = -2.0
propagador_foton = -1j / q2
factor_vertices = (-1j * e) ** 2
amplitud_esquematica = factor_vertices * propagador_foton

{'propagador_foton': propagador_foton, 'factor_vertices': factor_vertices, 'amplitud_esquematica': amplitud_esquematica}
```

## Paso 3. Lectura fisica

La celda anterior no incluye algebra de espinores ni indices de Lorentz, pero si fija el esqueleto de la amplitud. La dependencia en el momento transferido $q^2$ entra por la linea interna, mientras que la intensidad del acoplamiento electromagnetico entra por los vertices.

## Paso 4. Conexion con Ward e invariancia gauge

La consistencia gauge no es solo una condicion estetica. Se refleja en identidades como la de Ward, que controlan cancelaciones y aseguran que la descripcion perturbativa preserve la simetria local al nivel adecuado.
