# Problema resuelto 12: LSZ y amplitudes escalares

## Problema

Seguir de forma guiada como se pasa desde un correlador de campos escalares a una amplitud elemental de scattering mediante la idea de amputacion y reduccion LSZ.

## Paso 1. Correlador de cuatro puntos

Para un proceso escalar $2 \to 2$, un punto de partida natural es un correlador temporalmente ordenado de cuatro campos. Ese objeto contiene mas informacion que la amplitud final, porque incluye tambien la propagacion asociada a las patas externas.

## Paso 2. Estructura de polos y amputacion

La idea de LSZ es identificar los factores de polo asociados a las particulas externas, amputarlos y despues evaluar la amplitud sobre masa. El esquema conceptual se resume asi:

$$
\text{correlador completo} \longrightarrow \text{amputacion} \longrightarrow \mathcal{M}.
$$

Por eso una linea externa no debe tratarse igual que una interna.

```python
import numpy as np

lambda_coupling = 0.2
external_poles = [1 / 0.05, 1 / 0.04, 1 / 0.05, 1 / 0.04]
full_object = lambda_coupling * np.prod(external_poles)
amputated = full_object / np.prod(external_poles)

{'objeto_completo': full_object, 'objeto_amputado': amputated}
```

## Paso 3. Lectura fisica

La celda anterior no realiza una reduccion LSZ completa, pero sirve para fijar la idea: el correlador completo puede factorizarse en una parte universal asociada a patas externas y una parte interna que contiene la amplitud fisicamente relevante.

En una teoria escalar $\phi^4$, a nivel de arbol, esa amplitud elemental queda controlada por el acoplamiento $\lambda$.

## Paso 4. Conexion con diagramas

Una vez amputadas las patas externas, el diagrama a nivel de arbol se interpreta como representacion compacta del nucleo interno de la amplitud. Esto explica por que LSZ y reglas de Feynman no son temas separados, sino dos caras del mismo puente entre correladores y scattering.
