# Problema resuelto 07: Relatividad y campos

Notebook asociado a `Tutorial/02_relatividad_y_campos/README.md`.

## Problema

Justificar por que la mecanica cuantica de una sola particula no basta como teoria fundamental relativista.

## Resolucion

Porque la relatividad permite convertir energia en masa, lo que abre la posibilidad de crear nuevas particulas. Ademas, la interpretacion de ecuaciones relativistas de una sola particula presenta problemas con energias negativas y con la estructura probabilistica. El lenguaje de campos resuelve esto al admitir sectores de numero variable de excitaciones.
