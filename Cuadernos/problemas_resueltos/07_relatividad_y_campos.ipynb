# Problema resuelto 07: relatividad y campos

## Problema

Analizar por que la localidad de una teoria de campos ayuda a preservar causalidad relativista.


## Paso 1. Intervalo relativista

Dos eventos con separacion

$$
(x-y)^2 = (t_x - t_y)^2 - |\mathbf x - \mathbf y|^2
$$

son espaciolike si $(x-y)^2 < 0$. En ese caso no puede existir relacion causal directa entre ellos.


```python
def minkowski_interval(dt, dx, dy=0.0, dz=0.0):
    return dt**2 - dx**2 - dy**2 - dz**2

minkowski_interval(1.0, 2.0)
```

## Paso 2. Microcausalidad

Para observables bosonicos, la exigencia conceptual es

$$
[\mathcal O(x), \mathcal O(y)] = 0 \quad \text{si} \quad (x-y)^2 < 0.
$$

Esto asegura que experimentos suficientemente separados no se interfieran de forma instantanea.


## Paso 3. Papel de la localidad

Si el lagrangiano se construye localmente a partir de campos y derivadas en el mismo punto, la teoria tiene una oportunidad real de respetar la estructura causal relativista. Una interaccion manifiestamente no local mezclaria de entrada regiones alejadas del espacio-tiempo.


## Conclusion

La localidad no es solo un gusto estetico. Es una pista estructural fuerte de como una teoria cuantica puede compatibilizarse con relatividad especial sin permitir senales superluminales.
