# Problema resuelto 09: Cuantizacion del campo escalar

Notebook asociado a `Tutorial/04_cuantizacion_del_campo_escalar/README.md`.

## Problema

Explicar por que un campo libre puede verse como una coleccion infinita de osciladores armonicos cuanticos.

## Resolucion

La expansion en modos del campo separa la dinamica en componentes independientes etiquetadas por el momento. Cada una satisface una ecuacion equivalente a la del oscilador armonico. Al cuantizar cada modo aparecen operadores de creacion y aniquilacion, y el conjunto de todos ellos construye el espacio de Fock.
