# Problema resuelto 09: cuantizacion del campo escalar

## Problema

Explicar por que la cuantizacion de un campo libre conduce naturalmente a operadores de creacion y aniquilacion y al espacio de Fock.


## Paso 1. Campo clasico y modos normales

La ecuacion de Klein-Gordon admite soluciones en ondas planas. Al descomponer el campo en modos de momento, cada modo se comporta como un oscilador armonico independiente.

Ese es el punto de entrada para cuantizar: no se cuantiza solo una coordenada, sino una familia continua de amplitudes modales.


## Paso 2. Algebra de operadores

La promocion cuantica introduce operadores conmutadores

$$
[a_{\mathbf p}, a^\dagger_{\mathbf q}] = (2\pi)^3 \delta^{(3)}(\mathbf p - \mathbf q).
$$

Estos operadores permiten cambiar el numero de cuantos presentes en el estado.

## Paso 3. Interpretacion fisica

El vacio deja de ser simplemente la ausencia de movimiento clasico y pasa a ser el estado sin excitaciones del campo. Las particulas observadas son excitaciones cuantizadas sobre ese vacio.

La ventaja conceptual es enorme: la teoria describe sin esfuerzo procesos con numero variable de particulas.


## Conclusion

Cuantizar un campo libre equivale a cuantizar un continuo de osciladores. Esa observacion explica la aparicion de operadores de creacion y aniquilacion y justifica por que el espacio de Fock es el espacio de estados natural de la QFT perturbativa.
