# Ejemplo 09: bilineales y proyectores quirales

## Objetivo

Comparar de forma compacta algunos bilineales de Dirac y recordar como actuan los proyectores quirales sobre un espinor.

Los bilineales mas usados en el tutorial son

$$
\bar{\psi}\psi,
\qquad
\bar{\psi}\gamma^\mu\psi,
\qquad
\bar{\psi}\gamma^\mu\gamma^5\psi.
$$

Los proyectores quirales se escriben como

$$
P_L = \frac{1-\gamma^5}{2},
\qquad
P_R = \frac{1+\gamma^5}{2}.
$$


```python
import numpy as np

bilineales = {
    'escalar': 'psi_bar psi',
    'vector': 'psi_bar gamma^mu psi',
    'axial': 'psi_bar gamma^mu gamma^5 psi'
}

proyectores = {
    'P_L^2 = P_L': True,
    'P_R^2 = P_R': True,
    'P_L P_R = 0': True
}

bilineales, proyectores
```

## Lectura fisica

- El bilineal escalar aparece en terminos de masa de Dirac.
- El bilineal vectorial aparece en corrientes electromagneticas.
- El bilineal axial aparece en corrientes quirales y es especialmente importante para el sector debil.

Los proyectores permiten separar un espinor en sus componentes zurda y diestra, una operacion central para entender por que el Modelo Estandar es quiral.

```python
psi = np.array([1.0, -1.0, 2.0, 0.5])
PL_demo = np.array([1, 1, 0, 0])
PR_demo = np.array([0, 0, 1, 1])

{'psi_L_demo': psi * PL_demo, 'psi_R_demo': psi * PR_demo}
```

## Advertencia

La celda anterior solo ilustra la idea de descomponer componentes; no implementa matrices gamma reales. Su funcion es reforzar la intuicion de separacion quiral antes de volver a la formulacion algebraica completa.
