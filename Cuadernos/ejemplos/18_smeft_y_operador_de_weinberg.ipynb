# Ejemplo 18: SMEFT y operador de Weinberg

## Objetivo

Fijar la expansion de SMEFT y entender por que el operador de Weinberg de dimension cinco ocupa un lugar especial en la fisica de neutrinos.

## Resultado esperado

Ubicar el operador de Weinberg dentro de SMEFT y entender por que suprime masas de neutrinos con una escala alta.


La expansion efectiva se organiza como

$$
\mathcal{L}_{\mathrm{SMEFT}} = \mathcal{L}_{\mathrm{SM}} + \frac{1}{\Lambda}\sum c^{(5)}\mathcal{O}^{(5)} + \frac{1}{\Lambda^2}\sum c^{(6)}\mathcal{O}^{(6)} + \cdots.
$$

El operador de Weinberg aparece ya en dimension cinco, antes que la mayoria de las correcciones de nueva fisica discutidas en precision electrodébil.

```python
v = 246.0
Lambda = [1e5, 1e10, 1e14]
m_nu_demo = [round(v**2 / L, 12) for L in Lambda]
{'Lambda_GeV': Lambda, 'm_nu_demo_GeV': m_nu_demo}
```

## Lectura fisica

La cuenta anterior solo es dimensional, pero captura la intuicion mas importante: una escala efectiva muy alta puede traducirse en una masa neutrínica muy pequeña cuando el Higgs adquiere valor esperado. Esa es una de las razones por las que el operador de Weinberg es tan influyente.
