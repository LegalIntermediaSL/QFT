**Objetivo:** ilustrar con calculos concretos las razones por las que la mecanica cuantica de una particula es incompatible con la relatividad especial, y por que los campos cuanticos son la resolucion natural.

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

# Constantes en unidades naturales (hbar = c = 1)
m_e = 0.511  # masa del electron en MeV
```

## Ejemplo 1: Violacion de causalidad en MQ de una particula

En mecánica cuántica no relativista, el propagador libre es:

$$K(x, t; 0, 0) = \left(\frac{m}{2\pi i t}\right)^{3/2} e^{imx^2/(2t)}.$$

Este propagador es **no nulo para separaciones tipo espacio** ($|x| > ct$), lo que viola causalidad relativista. Ilustramos esto numericamente.

```python
# Propagador no relativista 1D (parte del modulo cuadrado)
def propagador_nr(x, t, m=1.0):
    """Modulo cuadrado del propagador libre no relativista 1D."""
    return m / (2 * np.pi * t)  # uniforme: |K|^2 = m/(2*pi*t)

t_vals = np.array([0.5, 1.0, 2.0])  # distintos tiempos
x_vals = np.linspace(0, 5, 300)

fig, ax = plt.subplots(figsize=(8, 4))
for t in t_vals:
    # El modulo cuadrado no depende de x en 1D: K es uniforme
    # Pero la fase si: mostramos Re(K) que tiene estructura espacial
    m = 1.0
    K_re = np.sqrt(m/(2*np.pi*t)) * np.cos(m*x_vals**2/(2*t))
    ax.plot(x_vals, K_re, label=f't = {t}')
    ax.axvline(t, linestyle='--', alpha=0.4)  # frente de luz: x = ct = t

ax.set_xlabel('x')
ax.set_ylabel('Re[K(x,t)]')
ax.set_title('Propagador libre no relativista: Re[K] es no nulo para x > t (linia discontinua)')
ax.legend()
plt.tight_layout()
plt.show()
print("El propagador es no nulo mas alla del frente de luz (lineas discontinuas).")
print("En QFT relativista, la amplitud exactamente se cancela para separaciones tipo espacio.")
```

## Ejemplo 2: Relacion de dispersion relativista y umbral de creacion de pares

En QFT, la energia de una particula libre satisface $E^2 = p^2 + m^2$. Para producir un par electron-positron en colision $e^-e^- \to e^-e^-e^+e^-$ se necesita una energia de umbral $E_{\text{umbral}} = 4m_e$ en el sistema del centro de masa.

```python
# Relacion de dispersion relativista
p_vals = np.linspace(0, 10*m_e, 500)  # momento en MeV

E_rel = np.sqrt(p_vals**2 + m_e**2)  # relativista
E_nr  = m_e + p_vals**2/(2*m_e)       # no relativista (aprox)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(p_vals/m_e, E_rel/m_e, 'b-', label='Relativista: $E=\\sqrt{p^2+m^2}$')
ax.plot(p_vals/m_e, E_nr/m_e,  'r--', label='No relativista: $E=m+p^2/2m$')
ax.axhline(4, color='g', linestyle=':', label='Umbral creacion de pares: $E=4m$')
ax.set_xlabel('p / m')
ax.set_ylabel('E / m')
ax.set_title('Relacion de dispersion: diferencia relativista vs no relativista')
ax.legend()
plt.tight_layout()
plt.show()

print(f"Umbral de creacion de pares e+e-: E_umbral = 4 * m_e = {4*m_e:.3f} MeV")
print("Por encima de este umbral, la MQ de particula fija no puede describir el proceso.")
print("QFT lo maneja naturalmente: el numero de particulas no es fijo.")
```

## Ejemplo 3: Localidad y conmutador causal

En QFT, los operadores de campo satisfacen:

$$[\phi(x), \phi(y)] = 0 \quad \text{para } (x-y)^2 < 0 \text{ (separacion tipo espacio)}.$$

Esto garantiza que mediciones en regiones causalmente desconectadas no se afectan. Ilustramos la estructura causal del espacio de Minkowski.

```python
fig, ax = plt.subplots(figsize=(6, 6))

# Cono de luz
x = np.linspace(-3, 3, 300)
ax.fill_between(x, x, 3, alpha=0.2, color='red', label='Futuro (tipo tiempo)')
ax.fill_between(x, -3, -np.abs(x), alpha=0.2, color='blue', label='Pasado (tipo tiempo)')
ax.fill_between(x, -np.abs(x), np.abs(x), alpha=0.1, color='green', label='Tipo espacio (conmutador = 0)')

ax.plot(x, np.abs(x), 'k-', linewidth=2)
ax.plot(x, -np.abs(x), 'k-', linewidth=2)
ax.plot(0, 0, 'ko', markersize=8)
ax.text(0.1, 0.1, 'Evento O', fontsize=10)

ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
ax.set_xlabel('x (posicion)')
ax.set_ylabel('t (tiempo)')
ax.set_title('Cono de luz y causalidad en QFT')
ax.legend(loc='upper right', fontsize=8)
plt.tight_layout()
plt.show()

print("Zona verde (tipo espacio): [phi(x), phi(y)] = 0")
print("=> Ningun observable en O puede afectar o ser afectado por eventos en la zona verde.")
print("   Esto es la version cuantica del principio de causalidad relativista.")
```

## Ejemplo 4: Por que necesitamos campos y no solo particulas

Comparacion de descripciones: particula vs campo.

```python
print("=" * 60)
print("COMPARACION: Mecanica Cuantica vs QFT")
print("=" * 60)

comparacion = [
    ("Descripcion",    "Estado cuantico de N particulas",  "Configuracion del campo phi(x)"),
    ("N particulas",   "Fijo (conservado)",                 "Variable (puede crearse/destruirse)"),
    ("Causalidad",     "Violada para separacion tipo espacio", "Garantizada: [phi(x),phi(y)]=0"),
    ("Antiparticulas", "Agregadas a mano (no naturales)",   "Emergen automaticamente del campo"),
    ("Espín",          "Postulado externo",                 "Consecuencia del grupo de Lorentz"),
    ("Interacciones",  "Potencial V(x1-x2)",                "Locales: L = L(phi, d_mu phi)"),
]

print(f"{'Aspecto':<20} {'MQ de particula':<40} {'QFT'}")
print("-" * 95)
for row in comparacion:
    print(f"{row[0]:<20} {row[1]:<40} {row[2]}")
```

## Resumen

| Problema de MQ | Resolucion en QFT |
|---|---|
| Propagador no nulo para tipo espacio | Conmutador de campos se anula |
| Numero de particulas fijo | Campo con operadores de creacion/aniquilacion |
| Creacion de pares sin descripcion | Excitaciones del campo |
| Causalidad violada | Localidad del Lagrangiano garantiza causalidad |
