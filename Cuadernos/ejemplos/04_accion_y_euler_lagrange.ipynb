# Ejemplo 04: accion y ecuaciones de Euler-Lagrange

## Objetivo

Ver de forma concreta como una densidad lagrangiana produce una ecuacion de campo. Esta es una de las rutas mas limpias para pasar del principio variacional al formalismo de QFT.


Partimos de la accion

$$
S[\phi] = \int d^4 x \, \mathcal L(\phi, \partial_\mu \phi).
$$

La condicion $\delta S = 0$ lleva a

$$
\frac{\partial \mathcal L}{\partial \phi} - \partial_\mu \left( \frac{\partial \mathcal L}{\partial (\partial_\mu \phi)} \right) = 0.
$$


```python
import sympy as sp

t, x, m = sp.symbols('t x m', positive=True)
phi = sp.Function('phi')(t, x)

L = sp.Rational(1, 2) * sp.diff(phi, t)**2 - sp.Rational(1, 2) * sp.diff(phi, x)**2 - sp.Rational(1, 2) * m**2 * phi**2
EL = sp.diff(L, phi) - sp.diff(sp.diff(L, sp.diff(phi, t)), t) - sp.diff(sp.diff(L, sp.diff(phi, x)), x)
sp.simplify(EL)
```

El resultado es equivalente a la ecuacion de Klein-Gordon en 1+1 dimensiones:

$$
\partial_t^2 \phi - \partial_x^2 \phi + m^2 \phi = 0.
$$

Lo importante aqui no es solo la ecuacion obtenida, sino el hecho de que la simetria y la dinamica quedan encapsuladas desde el inicio en la eleccion de $\mathcal L$.


## Comentario pedagogico

En teorias de gauge y en el Modelo Estandar seguimos la misma logica: escribimos la densidad lagrangiana mas general compatible con las simetrias y, a partir de ella, derivamos ecuaciones, corrientes conservadas, propagadores y vertices.
