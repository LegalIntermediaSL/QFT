# Ejemplo 16: QED, derivada covariante e identidad de Ward

## Objetivo

Reforzar de forma compacta la relacion entre simetria gauge local, derivada covariante, corriente electromagnetica e intuicion basica detras de la identidad de Ward.

## Resultado esperado

Identificar de forma compacta el paso entre simetria gauge, derivada covariante y consistencia de amplitudes en QED.


En QED, la derivada ordinaria se reemplaza por

$$
D_\mu = \partial_\mu + i e A_\mu,
$$

y el lagrangiano de Dirac pasa a acoplarse con el campo gauge de forma minima. La identidad de Ward resume que esta estructura gauge deja huellas directas en amplitudes y correladores.

```python
e = 0.303
A_mu = ['A0', 'A1', 'A2', 'A3']
D_mu = [f'd_mu + i {e} {a}' for a in A_mu]
{'derivada_covariante_componentes': D_mu}
```

## Lectura fisica

La celda anterior solo recuerda la forma del acoplamiento minimo, pero el mensaje importante es este: al promover la simetria global a local no solo aparece el campo $A_\mu$, sino que tambien queda fijada la forma del acoplamiento entre fermiones y fotones. La identidad de Ward expresa justamente la consistencia de esa estructura gauge cuando entramos en amplitudes.

```python
q_mu = [1.0, 0.0, 0.0, 1.0]
M_mu = [2.5, 0.3, -0.1, -2.5]
contraccion = sum(q*m for q, m in zip(q_mu, M_mu))
{'q_mu': q_mu, 'M_mu': M_mu, 'q_dot_M': contraccion}
```

## Advertencia

La contraccion anterior es solo una ilustracion algebraica. No sustituye una derivacion de la identidad de Ward a partir de invariancia gauge o del funcional generador, pero si ayuda a fijar la intuicion de que la amplitud no debe depender de grados de libertad gauge espurios.
