# Ejemplo 06: diagramas de Feynman basicos

## Objetivo

Introducir la lectura minima de un diagrama de Feynman como herramienta organizativa de la expansion perturbativa, no como dibujo literal de trayectorias clasicas.


## Regla pedagogica central

Un diagrama codifica terminos de una expansion de la matriz $S$. Cada linea interna representa un propagador y cada vertice representa un factor procedente del termino de interaccion en el lagrangiano.

Por ejemplo, para una teoria $\phi^4$ con

$$
\mathcal L_{\text{int}} = -\frac{\lambda}{4!} \phi^4,
$$

cada vertice aporta un factor proporcional a $-i\lambda$.


```python
def tree_level_order(num_vertices):
    return f'contribucion proporcional a lambda^{num_vertices}'

[tree_level_order(v) for v in range(1, 4)]
```

## Orden perturbativo

La utilidad inmediata del diagrama es indicar en que potencia del acoplamiento estamos trabajando. Un proceso con un vertice esta en orden $\lambda$, mientras que un diagrama con un bucle y dos vertices contribuye a orden $\lambda^2$.

Esto permite organizar calculos antes incluso de resolver integrales concretas.


## Advertencia importante

No debe leerse un diagrama de Feynman como si mostrara la historia clasica de una particula observada en el laboratorio. Es una notacion compacta para amplitudes. Las lineas internas corresponden a variables integradas, no a particulas detectables en un detector.
