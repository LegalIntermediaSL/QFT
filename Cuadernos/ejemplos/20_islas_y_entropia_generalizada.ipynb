# Ejemplo 20: islas y entropia generalizada

## Objetivo

Fijar la intuicion de entropia generalizada, tiempo de Page e inclusion de islas en la lectura moderna de la radiacion de Hawking.

## Resultado esperado

Entender por que la entropia generalizada y las islas refinan la lectura semiclasica de la curva de Page.


La idea central se resume en una expresion del tipo

$$
S_{\mathrm{gen}} = \frac{\mathrm{Area}}{4G_N} + S_{\mathrm{bulk}}.
$$

La novedad conceptual es que la entropia fina relevante no siempre coincide con la de la radiacion exterior tratada de forma aislada.

```python
area_demo = [120.0, 90.0, 60.0]
S_bulk_demo = [5.0, 18.0, 26.0]
S_gen_demo = [a/4 + s for a, s in zip(area_demo, S_bulk_demo)]
{'area_demo': area_demo, 'S_bulk_demo': S_bulk_demo, 'S_gen_demo': S_gen_demo}
```

## Lectura fisica

La celda anterior no modela un agujero negro real, pero si ayuda a ver la logica: distintas contribuciones compiten en el calculo de la entropia generalizada y, en ciertos regimenes, una configuracion con isla puede volverse la descripcion correcta de la entropia fina.

```python
t = [0, 1, 2, 3, 4]
S_sin_isla = [0, 2, 4, 6, 8]
S_con_isla = [0, 2, 4, 3, 2]
{'t_demo': t, 'sin_isla': S_sin_isla, 'con_isla': S_con_isla}
```

## Advertencia

Estas celdas son solo modelos pedagogicos. No sustituyen una derivacion con replica trick, superficies extremales o wormholes. El objetivo es conservar la intuicion de por que la curva de Page cambia cuando la entropia se calcula de forma mas refinada.
