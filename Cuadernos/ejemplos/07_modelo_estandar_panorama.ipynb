# Ejemplo 07: panorama del Modelo Estandar

## Objetivo

Ordenar la arquitectura del Modelo Estandar desde el punto de vista lagrangiano: sector gauge, sector fermionico, Higgs y Yukawa.


El grupo de gauge es

$$
SU(3)_c \times SU(2)_L \times U(1)_Y.
$$

A partir de el se organizan las interacciones fuerte y electrodebil. La gravedad queda fuera del alcance del Modelo Estandar.


```python
sectores = {
    'gauge': ['gluones', 'W', 'Z', 'foton'],
    'fermiones': ['quarks', 'leptones'],
    'Higgs': ['doblete escalar'],
    'Yukawa': ['masas fermionicas y mezclas'],
}
sectores
```

## Esquema sintetico

$$
\mathcal L_{\text{SM}} = \mathcal L_{\text{gauge}} + \mathcal L_{\text{ferm}} + \mathcal L_{\text{Higgs}} + \mathcal L_{\text{Yukawa}}.
$$

Cada bloque responde a una pregunta distinta:

- como se propagan los bosones de gauge;
- como se acoplan quarks y leptones;
- como se rompe espontaneamente la simetria electrodebil;
- como adquieren masa los fermiones.


## Observacion clave

El Modelo Estandar es extraordinariamente exitoso y, sin embargo, deja preguntas abiertas: masas de neutrinos, materia oscura, jerarquia, gravedad cuantica y el origen mas profundo de sus parametros libres.
