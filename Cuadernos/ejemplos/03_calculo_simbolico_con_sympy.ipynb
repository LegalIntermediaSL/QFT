# Cálculo Simbólico en QFT con SymPy

Este notebook demuestra cómo utilizar la librería **SymPy** para realizar manipulaciones algebraicas comunes en Teoría Cuántica de Campos, como el álgebra de Clifford y el cálculo de trazas de matrices Gamma.

```python
from sympy import symbols, init_printing
from sympy.physics.hep.gamma_matrices import GammaMatrix as G, gamma_trace
from sympy.physics.quantum.dagger import Dagger

init_printing(use_latex='mathjax')
print("Librerías cargadas correctamente.")
```

## 1. Verificación del Álgebra de Clifford

El álgebra de Clifford define la relación fundamental entre las matrices $\gamma^\mu$:

$$\{\gamma^\mu, \gamma^\nu\} = 2\eta^{\mu\nu}$$

En SymPy, estas matrices están predefinidas.

```python
mu, nu = symbols('mu nu', dummy=True)

# Ejemplo: Producto de dos matrices gamma
expr = G(mu) * G(nu)
expr
```

## 2. Cálculo de Trazas

Las trazas de productos de matrices gamma son esenciales para calcular amplitudes de dispersión. 

- $\text{Tr}(\gamma^\mu) = 0$
- $\text{Tr}(\gamma^\mu\gamma^\nu) = 4\eta^{\mu\nu}$

```python
# Traza de dos matrices gamma
trace_2 = gamma_trace(G(mu) * G(nu))
print(f"Tr(gamma^mu * gamma^nu) = {trace_2}")

# Traza de cuatro matrices gamma
rho, sigma = symbols('rho sigma', dummy=True)
trace_4 = gamma_trace(G(mu) * G(nu) * G(rho) * G(sigma))
trace_4
```

## 3. Conclusión

SymPy permite automatizar cálculos que manualmente suelen ser propensos a errores, especialmente cuando el número de matrices gamma aumenta (por ejemplo, en procesos con varios lazos o múltiples partículas externas).
