# Ejemplo 19: matching UV a SMEFT

## Objetivo

Introducir de forma compacta la idea de matching entre una teoria UV y su EFT, y fijar el papel de los coeficientes de Wilson.

## Resultado esperado

Fijar la idea de matching y el papel de los coeficientes de Wilson como huella IR de una teoria UV.


La logica de matching es simple:

1. se calcula una amplitud o correlador en la teoria UV;
2. se calcula la misma cantidad en la EFT;
3. se ajustan los coeficientes de Wilson para que ambas coincidan a baja energia.


```python
g = 0.8
M = 2000.0
C6_demo = g**2 / M**2
{'g_demo': g, 'M_demo_GeV': M, 'C6_demo_GeV^-2': C6_demo}
```

## Lectura fisica

La cuenta anterior reproduce la intuicion mas basica: al integrar un mediador pesado, el coeficiente del operador efectivo hereda el tamano del acoplamiento UV y la supresion por la masa pesada. Ese numero es justamente una version simplificada de un coeficiente de Wilson.

```python
mu = [2000.0, 500.0, 100.0]
C_demo_running = [round(C6_demo * (1 + 0.05*i), 10) for i, _ in enumerate(mu)]
{'mu_GeV': mu, 'C_demo_running': C_demo_running}
```

## Advertencia

Estas celdas no hacen matching real ni running renormalizado completo. Solo fijan la cadena conceptual: primero matching en la escala pesada, luego evolucion efectiva hacia escalas mas bajas.
