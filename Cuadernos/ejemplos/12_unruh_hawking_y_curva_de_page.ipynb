# Ejemplo 12: Unruh, Hawking y curva de Page

## Objetivo

Reforzar la intuicion de que acceso parcial, horizontes y mezclas de modos pueden producir termicidad efectiva, y conectar esa idea con la curva de Page como organizador conceptual de la tension entre termicidad y unitaridad.

## Resultado esperado

Conservar una intuicion compacta de como acceso parcial, termicidad efectiva y curva de Page se encadenan en la paradoja de la informacion.


El efecto Unruh y la radiacion de Hawking comparten una arquitectura conceptual:

- un observador no accede a todos los grados de libertad;
- el estado reducido accesible puede parecer termico;
- la pregunta por la informacion obliga a pensar en unitaridad global y subsistemas.

```python
import numpy as np

tiempo = np.linspace(0.0, 1.0, 9)
page_demo = np.where(tiempo <= 0.5, 2 * tiempo, 2 * (1 - tiempo))
list(zip(np.round(tiempo, 2), np.round(page_demo, 3)))
```

## Lectura fisica

La celda anterior no calcula una curva de Page real. Solo dibuja un perfil de subida y bajada para fijar la intuicion de que la entropia de radiacion no deberia crecer para siempre si la evolucion global es unitaria.

Ese cambio de comportamiento es justamente lo que vuelve tan profunda la paradoja de la informacion.

```python
aceleracion = np.array([0.5, 1.0, 2.0, 4.0])
temperatura_unruh_demo = aceleracion / (2 * np.pi)
list(zip(aceleracion.tolist(), np.round(temperatura_unruh_demo, 4).tolist()))
```

## Advertencia

Las celdas numericas son solo modelos de intuicion. No sustituyen ni la derivacion de modos de Rindler ni el calculo de Hawking, pero ayudan a conectar tres ideas: termicidad efectiva, acceso parcial a informacion y evolucion de la entropia de radiacion.
