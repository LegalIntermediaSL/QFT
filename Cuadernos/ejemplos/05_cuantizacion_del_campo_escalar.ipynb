# Ejemplo 05: cuantizacion del campo escalar

## Objetivo

Traducir la intuicion del oscilador armonico al lenguaje del campo escalar libre y del espacio de Fock.


La expansion del campo libre se escribe como

$$
\phi(x) = \int \frac{d^3 p}{(2\pi)^3} \frac{1}{\sqrt{2 E_{\mathbf p}}}
\left(a_{\mathbf p} e^{-i p \cdot x} + a^\dagger_{\mathbf p} e^{i p \cdot x}\right),
$$

con $E_{\mathbf p} = \sqrt{\mathbf p^2 + m^2}$. Los operadores satisfacen

$$
[a_{\mathbf p}, a^\dagger_{\mathbf q}] = (2\pi)^3 \delta^{(3)}(\mathbf p - \mathbf q).
$$


```python
import numpy as np

m = 1.0
p = np.array([0.0, 0.5, 1.0, 2.0])
E = np.sqrt(p**2 + m**2)
list(zip(p, E))
```

## Interpretacion

El estado vacio $|0\rangle$ satisface $a_{\mathbf p}|0\rangle = 0$ para todo $\mathbf p$. Un estado de una particula se construye como

$$
|\mathbf p\rangle = a^\dagger_{\mathbf p} |0\rangle.
$$

El espacio de Hilbert adecuado deja de ser el de una sola particula y pasa a ser el espacio de Fock, que acomoda cero, una, dos o muchas excitaciones.
