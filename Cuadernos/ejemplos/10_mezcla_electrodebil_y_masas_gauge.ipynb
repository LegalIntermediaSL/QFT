# Ejemplo 10: mezcla electrodébil y masas gauge

## Objetivo

Seguir de forma compacta como la mezcla entre $W^3$ y $B$ produce el foton y el boson $Z$, y como el vacio del Higgs fija las masas de $W$ y $Z$.

En el sector electrodébil, las combinaciones fisicas neutras se escriben como

$$
A_\mu = B_\mu \cos\theta_W + W^3_\mu \sin\theta_W,
\qquad
Z_\mu = -B_\mu \sin\theta_W + W^3_\mu \cos\theta_W.
$$

Ademas, al expandir el termino cinetico del Higgs alrededor del vacio aparecen las relaciones estructurales para $m_W$ y $m_Z$.

```python
import numpy as np

g = 0.65
gp = 0.36
v = 246.0

theta_w = np.arctan(gp / g)
m_w = g * v / 2.0
m_z = v * np.sqrt(g**2 + gp**2) / 2.0

{'theta_W_rad': theta_w, 'sin_theta_W': np.sin(theta_w), 'm_W_GeV': m_w, 'm_Z_GeV': m_z, 'mZ_over_mW': m_z / m_w}
```

## Lectura fisica

La mezcla electrodébil muestra que el foton no estaba aislado desde el principio como campo gauge elemental. Surge como combinacion no rota seleccionada por el vacio. La misma reorganizacion explica por que el boson $Z$ y los bosones $W^\pm$ adquieren masa mientras el foton permanece sin masa.

```python
W3 = np.array([0.0, 1.0])
B = np.array([1.0, 0.0])
A = np.cos(theta_w) * B + np.sin(theta_w) * W3
Z = -np.sin(theta_w) * B + np.cos(theta_w) * W3

{'A_demo': np.round(A, 4), 'Z_demo': np.round(Z, 4)}
```

## Advertencia

La ultima celda solo ilustra la idea de combinacion lineal en un espacio de dos componentes; no es una diagonalizacion completa del lagrangiano. Su funcion es fijar la intuicion de mezcla antes de volver a la formulacion formal.
