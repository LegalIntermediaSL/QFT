# El Potencial de Higgs y el Sombrero Mexicano

Este notebook visualiza el potencial escalar responsable de la ruptura espontánea de la simetría en el Modelo Estándar.

El potencial tiene la forma:
$$V(\phi) = \mu^2 |\phi|^2 + \lambda |\phi|^4$$

- Si $\mu^2 > 0$: El mínimo está en el origen (simetría preservada).
- Si $\mu^2 < 0$: El origen se vuelve un máximo local y aparece un conjunto de mínimos en un círculo (ruptura espontánea).

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def higgs_potential(phi_re, phi_im, mu2, lam):
    phi2 = phi_re**2 + phi_im**2
    return mu2 * phi2 + lam * phi2**2

print("Funciones cargadas.")
```

## Visualización 3D: Ruptura de Simetría

Configuramos $\mu^2 < 0$ y $\lambda > 0$.

```python
re = np.linspace(-2, 2, 100)
im = np.linspace(-2, 2, 100)
RE, IM = np.meshgrid(re, im)

# Parámetros para el Sombrero Mexicano
mu2 = -2
lam = 0.5
Z = higgs_potential(RE, IM, mu2, lam)

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(RE, IM, Z, cmap='magma', edgecolor='none', alpha=0.8)
ax.set_title("Potencial de Higgs: Ruptura Espontánea de Simetría")
ax.set_xlabel("Re(φ)")
ax.set_ylabel("Im(φ)")
ax.set_zlabel("V(φ)")
plt.show()
```
