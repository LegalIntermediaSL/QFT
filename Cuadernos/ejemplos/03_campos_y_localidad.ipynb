# Ejemplo 03: campos y localidad

## Objetivo

Ilustrar por que un campo local es una mejor variable fundamental que una descripcion basada en particulas puntuales con numero fijo.


## Del oscilador al campo

Un campo clasico escalar libre satisface la ecuacion de Klein-Gordon

$$
(\Box + m^2) \phi(x) = 0.
$$

Su descomposicion en modos convierte el problema en una familia continua de osciladores. Lo realmente nuevo es que esos modos estan distribuidos sobre todo el espacio y permiten formular interacciones locales.


```python
import numpy as np

x = np.linspace(-6, 6, 200)
phi = np.exp(-x**2 / 2.0) * np.cos(2 * x)

phi[:5]
```

La funcion anterior puede verse como un paquete de onda simplificado. En una teoria de campos, la informacion relevante no es la posicion exacta de una particula clasica, sino la amplitud del campo en cada punto.


## Localidad

Una densidad lagrangiana local tipica depende de los campos y de sus derivadas en el mismo punto $x$:

$$
\mathcal L(x) = \frac{1}{2} \partial_\mu \phi \, \partial^\mu \phi - \frac{1}{2} m^2 \phi^2 - V(\phi).
$$

No se mezclan directamente valores del campo en puntos alejados. Esta exigencia ayuda a preservar causalidad y a implementar el principio de descomposicion de cluster.
