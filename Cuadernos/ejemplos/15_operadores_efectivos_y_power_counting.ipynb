# Ejemplo 15: operadores efectivos y power counting

## Objetivo

Fijar la relacion entre dimension del operador, escala de corte $\Lambda$ y tamano esperado de una correccion efectiva a baja energia.

## Resultado esperado

Reconocer como la dimension del operador y la escala de corte organizan el tamano esperado de correcciones efectivas.


En una EFT tipica escribimos

$$
\mathcal{L}_{\mathrm{eff}} = \mathcal{L}_{d\leq 4} + \sum_{n>4}\frac{c_n}{\Lambda^{n-4}}\mathcal{O}_n.
$$

La idea central es que los operadores de mayor dimension quedan suprimidos por potencias de $E/\Lambda$.

```python
import numpy as np

E = np.array([1, 5, 10, 50], dtype=float)
Lambda = 1000.0
sup_dim6 = np.round((E / Lambda) ** 2, 8)
sup_dim8 = np.round((E / Lambda) ** 4, 12)

{
    'E_GeV': E.tolist(),
    'dim6': sup_dim6.tolist(),
    'dim8': sup_dim8.tolist()
}
```

## Lectura fisica

La celda anterior muestra una idea muy simple pero central: si la escala nueva esta muy por encima de la energia del experimento, los operadores de dimension seis y ocho quedan rapidamente suprimidos. Esa es la razon por la que una EFT puede ser predictiva incluso sin conocer toda la UV completion.

```python
dimensiones = {'phi': 1.0, 'psi': 1.5, 'F_munu': 2.0}
operadores = {
    'phi^4': 4.0,
    '(psibar gamma psi)^2': 6.0,
    '(F^2)^2': 8.0
}
clasificacion = {
    nombre: ('relevante' if d < 4 else 'marginal' if d == 4 else 'irrelevante')
    for nombre, d in operadores.items()
}
{'dimensiones': dimensiones, 'clasificacion': clasificacion}
```
