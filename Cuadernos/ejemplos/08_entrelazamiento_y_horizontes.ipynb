# Ejemplo 08: entrelazamiento y horizontes

## Objetivo

Abrir el puente entre QFT, informacion cuantica y agujeros negros. El foco no es tecnico-exhaustivo, sino mostrar por que el vacio cuantico y la particion del espacio-tiempo producen entropia de entrelazamiento.


## Idea central

En QFT, el vacio no es un producto trivial de estados independientes en regiones espaciales separadas. Si dividimos el sistema en una region $A$ y su complemento, el estado reducido

$$
\rho_A = \mathrm{Tr}_{\bar A} \, \rho
$$

tiene entropia de von Neumann

$$
S_A = - \mathrm{Tr}(\rho_A \log \rho_A).
$$

Esa entropia mide entrelazamiento entre grados de libertad a ambos lados de la frontera.


```python
import numpy as np

eigenvalues = np.array([0.5, 0.5])
entropy = -np.sum(eigenvalues * np.log(eigenvalues))
entropy
```

Este ejemplo binario no es una QFT real, pero recuerda la logica del calculo: la informacion accesible a un observador parcial se describe mediante una matriz de densidad reducida, no mediante un estado puro global.


## Horizonte y termicidad efectiva

Cuando un horizonte impide acceder a parte de los grados de libertad, el estado reducido puede parecer termico. Esta idea aparece en el efecto Unruh y en la radiacion de Hawking.
