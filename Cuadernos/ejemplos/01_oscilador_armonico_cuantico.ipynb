# Ejemplo 01: Oscilador armonico cuantico

Este cuaderno queda preparado como base para ejemplos relacionados con el oscilador armonico cuantico, operadores de creacion y aniquilacion, y su relacion con la cuantizacion de campos.

## Ideas para desarrollar

- construir el hamiltoniano en forma algebraica;
- verificar conmutadores;
- generar estados de numero;
- conectar con modos de un campo libre.
