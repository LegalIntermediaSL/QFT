# Ejemplo 01: oscilador armonico cuantico

## Objetivo

Conectar el oscilador armonico cuantico con la cuantizacion de campos. El oscilador es el modelo elemental que reaparece una y otra vez en QFT cuando descomponemos un campo libre en modos normales.

## Prerrequisitos

- algebra lineal basica;
- operadores y conmutadores;
- energia del oscilador clasico y cuantico.


## Idea fisica

En mecanica cuantica el Hamiltoniano del oscilador es

$$
\hat H = \hbar \omega \left( \hat a^\dagger \hat a + \frac{1}{2} \right).
$$

En QFT, despues de expandir el campo en modos de Fourier, cada modo de momento se comporta como un oscilador armonico independiente. Por eso entender bien este sistema no es un lujo pedagogico: es una puerta de entrada al espacio de Fock.


```python
import numpy as np

hbar = 1.0
omega = 1.5
n_values = np.arange(6)
energies = hbar * omega * (n_values + 0.5)

list(zip(n_values, energies))
```

## Resultado

Los niveles aparecen igualmente espaciados:

$$
E_n = \hbar \omega \left(n + \frac{1}{2}\right), \qquad n = 0, 1, 2, \dots
$$

La separacion uniforme entre niveles es exactamente lo que hace posible interpretar $\hat a^\dagger$ como creador de cuantos y $\hat a$ como aniquilador.


```python
level_spacing = np.diff(energies)
level_spacing
```

## Puente con QFT

Cuando un campo escalar libre se escribe como superposicion de modos,

$$
\phi(x) = \int \frac{d^3 p}{(2\pi)^3} \frac{1}{\sqrt{2 E_{\mathbf p}}} \left(a_{\mathbf p} e^{-i p \cdot x} + a^\dagger_{\mathbf p} e^{i p \cdot x} \right),
$$

cada par $(a_{\mathbf p}, a^\dagger_{\mathbf p})$ reproduce la estructura algebraica del oscilador. La gran diferencia es que ahora tenemos un continuo de modos etiquetados por el momento.
