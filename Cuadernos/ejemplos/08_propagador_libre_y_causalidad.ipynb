# Ejemplo 08: propagador libre y causalidad

## Objetivo

Ilustrar por que el propagador escalar libre es una funcion de Green central en QFT, como aparece la prescripcion $i\epsilon$ y por que no debe confundirse el propagador de Feynman con el conmutador causal.

Para un campo escalar libre, el propagador de Feynman en espacio de momentos se escribe como

$$
\tilde{\Delta}_F(p) = \frac{i}{p^2 - m^2 + i\epsilon}.
$$

Este objeto reaparece en correladores, lineas internas de diagramas y funciones de Green del operador de Klein-Gordon. La microcausalidad, sin embargo, se expresa de manera mas directa a traves del conmutador local $[\phi(x),\phi(y)]$.

```python
import numpy as np

m = 1.0
epsilon = 1e-2
p0 = np.linspace(-2.5, 2.5, 11)
p = 1.2

denominator = p0**2 - p**2 - m**2 + 1j * epsilon
propagator = 1j / denominator

list(zip(np.round(p0, 2), np.round(propagator.real, 4), np.round(propagator.imag, 4)))
```

## Lectura fisica

La tabla anterior no es una observacion experimental directa, sino una manera de inspeccionar la estructura analitica del propagador. Los polos aparecen cuando el denominador se acerca a la condicion on-shell.

La prescripcion $i\epsilon$ desplaza esos polos y fija la eleccion consistente del contorno al integrar en energia.

```python
x = np.linspace(-4.0, 4.0, 9)
commutator_model = np.sign(x) * np.exp(-np.abs(x))

list(zip(np.round(x, 2), np.round(commutator_model, 4)))
```

## Advertencia conceptual

La celda anterior no calcula el conmutador relativista exacto; solo da una imagen sencilla de un objeto antisimetrico para recordar la idea de que el **conmutador** y el **propagador temporalmente ordenado** no son el mismo objeto.

En el tutorial, esta distincion se vuelve crucial para entender por que:

- el propagador de Feynman es util para calculo perturbativo;
- la microcausalidad se formula con conmutadores locales;
- una linea interna de Feynman no debe leerse como trayectoria clasica.
