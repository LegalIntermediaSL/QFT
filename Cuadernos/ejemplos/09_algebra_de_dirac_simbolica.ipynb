## 1. La Matrix $\gamma^5$

La matriz $\gamma^5$ es fundamental para definir la quiralidad:
$$\gamma^5 = i\gamma^0\gamma^1\gamma^2\gamma^3$$

Propiedades matemáticas:
- $(\gamma^5)^2 = 1$
- $\{\gamma^5, \gamma^\mu\} = 0$

```python
# En SymPy, G(5) representa la matriz gamma^5
gamma5 = G(5)
mu = symbols('mu', dummy=True)

# Traza de gamma^5 con 4 matrices gamma (tensor de Levi-Civita)
nu, rho, sigma = symbols('nu rho sigma', dummy=True)
expr = G(mu) * G(nu) * G(rho) * G(sigma) * G(5)
result = gamma_trace(expr)
print("Resultado de Tr(gamma^mu gamma^nu gamma^rho gamma^sigma gamma^5):")
result
```
