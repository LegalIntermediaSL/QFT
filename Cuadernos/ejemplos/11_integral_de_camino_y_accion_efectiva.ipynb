# Ejemplo 11: integral de camino y accion efectiva

## Objetivo

Fijar la cadena conceptual que conecta integral funcional, funcional generador, accion efectiva y potencial efectivo en una teoria escalar sencilla.

El formalismo funcional organiza la teoria alrededor de objetos como

$$
Z[J], \qquad W[J] = -i \log Z[J], \qquad \Gamma[\phi_c].
$$

La accion efectiva y el potencial efectivo permiten estudiar no solo amplitudes, sino tambien la estructura del vacio cuando se incorporan fluctuaciones cuanticas.

```python
import numpy as np

phi = np.linspace(-2.0, 2.0, 9)
m2 = -1.0
lam = 0.4
V_classico = 0.5 * m2 * phi**2 + 0.25 * lam * phi**4
correccion = 0.05 * np.log(1.0 + phi**2)
V_efectivo_demo = V_classico + correccion

list(zip(np.round(phi, 2), np.round(V_classico, 4), np.round(V_efectivo_demo, 4)))
```

## Lectura fisica

La celda anterior no calcula un potencial efectivo realista a un lazo, pero si ayuda a visualizar la idea esencial: al incorporar correcciones cuanticas, la forma del potencial puede desplazarse o deformarse.

Eso basta para entender por que la accion efectiva es tan importante para vacios, ruptura espontanea de simetria y estabilidad.

```python
phi_c_demo = np.array([-1.2, 0.0, 1.2])
Gamma_demo = np.round(0.5 * phi_c_demo**2 + 0.1 * phi_c_demo**4, 4)
{'phi_c_demo': phi_c_demo.tolist(), 'Gamma_demo': Gamma_demo.tolist()}
```

## Advertencia

Las dos celdas numericas son solo modelos pedagogicos para fijar intuicion. No sustituyen una derivacion funcional completa de $\Gamma[\phi_c]$ ni de $V_{\text{ef}}$, pero ayudan a ver que la teoria cuantica puede reorganizar la estructura del minimo.
